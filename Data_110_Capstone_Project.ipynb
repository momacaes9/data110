```python

import pandas as pd

from google.colab import drive
drive.mount('/content/drive')

data_path = '/content/drive/My Drive/Copy of Bank_Account_Fraud.csv'
df = pd.read_csv(data_path)

df.info()
df.head()
```

```python
#Graph made with velocity_6h as the numerical y axis variable with the fraud_bool as the categorical x axis variable.
#This graph does include outliers
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(data=df, x='fraud_bool', y='velocity_6h', color= "pink")

plt.title('Application Velocity vs. Bank Account Fraud')
plt.xlabel('Fraud Status [0 = No Fraud, 1 = Fraud]')
plt.ylabel('Application Velocity (6 Hours)')
plt.ylim(0, 18000) #Y axis range
plt.show()
```

```python
#This graph does not include outliers for a cleaner data visual BUT should not be used for data analysis.
import seaborn as sns
import matplotlib.pyplot as plt

sns.boxplot(data=df, x='fraud_bool', y='velocity_6h', color= "hotpink",  showfliers=False)

plt.title('Application Velocity vs. Bank Account Fraud')
plt.xlabel('Fraud Status [0 = No Fraud, 1 = Fraud]')
plt.ylabel('Application Velocity (6 Hours)')
plt.ylim(0, 18000) #Y axis range
plt.show()
```
